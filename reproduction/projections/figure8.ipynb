```python
import subprocess, sys
from pathlib import Path

ROOT = Path.cwd().parents[1]
print("root", ROOT)
```

```python
# Figure 8 is produced by a script, not by notebook code. Running it here keeps
# the reproduction path in one place.
#
# It writes, into paper/Figures/Projections/:
#   projections_cmip6.pdf            the 2x3 regional panel, SSP5-8.5
#   projections_cmip6_scenarios.pdf  the global aggregate across the three SSPs
#   projections_cmip6_regions_2100*.csv and the per-country impact tables

cmd = [sys.executable, "scripts/2026-09-28-projections-cmip6.py"]
print(" ".join(cmd))
r = subprocess.run(cmd, cwd=ROOT, capture_output=True, text=True)
print(r.stdout[-4000:])
if r.returncode:
    print("STDERR\n", r.stderr[-3000:])
```
