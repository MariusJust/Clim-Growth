## 1. Data preparation

Two things differ from `global_model.ipynb` and both matter.

**The network takes three inputs**, `(T, P, t)`, not two. The grid has to carry a time
column.

**`create_pred_input(time_periods=...)` cannot be used here.** It emits a raw
`arange(0, time_periods+1)` for the time column, but this run was trained with
`time_scaling: standardize`, which maps `1..T` onto mean 0 and unit standard deviation
using the deterministic moments `mu = (T+1)/2` and `sigma = sqrt((T^2-1)/12)`. Feeding
raw indices would evaluate the network far outside the range it saw. We therefore build
the grid here with the same transform the training code applies.


```python
data = pd.read_excel(ROOT / "data/MainData.xlsx")
if m.get("data_end"):
    data = data[data.Year <= m["data_end"]]

Tp = data.pivot(index="Year", columns="CountryCode", values="TempPopWeight")
Pv = data.pivot(index="Year", columns="CountryCode",
                values="PrecipPopWeight").reindex_like(Tp).to_numpy(float)
Tv = Tp.to_numpy(float)
years = np.asarray(Tp.index, dtype=int)

mean_temp,   std_temp   = float(np.nanmean(Tv)), float(np.nanstd(Tv))
mean_precip, std_precip = float(np.nanmean(Pv)), float(np.nanstd(Pv))

n_periods = len(years)
mean_t, std_t = (n_periods + 1) / 2, np.sqrt((n_periods**2 - 1) / 12)

# Temperature axis restricted to the 5th-95th percentile of the sample: only 3 of
# 10,324 country-years exceed 30 C, so the tails are functional form, not evidence.
t_lo, t_hi = np.nanpercentile(Tv, [5, 95])
precip_med = float(np.nanmedian(Pv))

print(f"panel {n_periods} x {Tv.shape[1]}   years {years[0]}-{years[-1]}")
print(f"temperature {mean_temp:.3f} +/- {std_temp:.3f} C, plotted over {t_lo:.1f}-{t_hi:.1f} C")
print(f"precipitation held at the sample median, {precip_med:.0f} mm")
```

## 3. Prediction on the temperature-time grid

The forward pass is done in NumPy rather than through `model_visual.predict`. It
reproduces the Keras graph exactly (hidden layers are `Dense` with a bias and a swish
activation, the output layer is a bias-free `Dense` of width 1) and it needs neither
TensorFlow nor a rebuilt model object, so the notebook runs anywhere. It has been
checked against the recorded BIC of the saved runs.

Each surface is **centred at the sample-mean temperature within each year**. That
removes the common time level, which in this specification is confounded with the year
effects the model does not carry, and leaves the shape of the temperature response.


```python
def swish(z):
    return z / (1.0 + np.exp(-z))


def forward(w, X):
    """Chained Dense stack: hidden layers carry a bias, the output layer does not."""
    idx = sorted({int(k.split("/")[1].split("_")[1]) if "_" in k.split("/")[1] else 0
                  for k in w if k.startswith("layers")})
    h = X
    for i in idx:
        b = f"layers/dense{'' if i == 0 else f'_{i}'}/vars"
        h = h @ np.asarray(w[f"{b}/0"], float)
        if f"{b}/1" in w:
            h = swish(h + np.asarray(w[f"{b}/1"], float))
    return h[..., 0]


def surface(node, n_temp=120):
    """Centred response in pp/yr on a (temperature, year) grid."""
    w = read_weights(str(run_dir / "parameters" / f"{node}.weights.h5"))
    Tg = np.linspace(t_lo, t_hi, n_temp)
    TT, tt = np.meshgrid(Tg, np.arange(1, n_periods + 1), indexing="ij")
    X = np.stack([(TT - mean_temp) / std_temp,
                  np.full_like(TT, (precip_med - mean_precip) / std_precip),
                  (tt - mean_t) / std_t], axis=-1)
    G = forward(w, X)
    return Tg, (G - G.mean(axis=0, keepdims=True)) * 100.0


Tg, G = surface(top_models[0])
print(f"BIC-best {top_models[0]}: surface {G.shape}  range {G.min():.2f} to {G.max():.2f} pp")
```

## 5. Is the surface time-varying?

Decompose the centred surface `g(T, t)` into a time-invariant shape and a remainder:

`g(T, t) = gbar(T) + r(T, t)`

and report `||r||^2 / ||g||^2`, the share of surface variation that is time-varying. A
value near zero means the response is effectively static; near one means no stable shape
survives. `corr(first, last)` compares the response shape in the first and last year: near
+1 means the shape is preserved, near -1 means it has inverted.


```python
rows = []
for node in top_models:
    if not (run_dir / "parameters" / f"{node}.weights.h5").exists():
        continue
    _, Gi = surface(node)
    gbar = Gi.mean(axis=1, keepdims=True)
    r = Gi - gbar
    rows.append(dict(
        arch=str(node),
        BIC=results[node][0],
        time_varying_pct=(r**2).sum() / (Gi**2).sum() * 100,
        corr_first_last=np.corrcoef(Gi[:, 0], Gi[:, -1])[0, 1],
        invariant_amp_pp=gbar.max() - gbar.min(),
        amp_min_pp=(Gi.max(axis=0) - Gi.min(axis=0)).min(),
        amp_max_pp=(Gi.max(axis=0) - Gi.min(axis=0)).max(),
    ))

summary = pd.DataFrame(rows).sort_values("BIC").reset_index(drop=True)
summary.round(3)
```

**Read the spread, not the first row.** If the time-varying share and the sign of
`corr(first, last)` move substantially across architectures that sit within a few tens of
BIC of one another, the time variation is not identified: models a reader would regard as
equally good disagree about whether the response changes over time. On the 2026-09-08 run
the share ranges from about 4 percent to about 92 percent and the correlation changes
sign, which is why PROJECT.md decision D17 blocks any surface-shape claim until the
multi-init study reports.


```python
fig_lines = go.Figure()
for j, yr in enumerate(years):
    if yr % 10 != 0 and yr not in (years[0], years[-1]):
        continue
    fig_lines.add_trace(go.Scatter(x=Tg, y=G[:, j], mode="lines", name=str(yr)))

fig_lines.update_layout(
    xaxis_title="Temperature (°C)",
    yaxis_title="Growth response (pp/yr)",
    legend_title="Year",
    font=dict(size=11),
    margin=dict(l=40, r=10, b=40, t=20),
)
fig_lines.show()
```

## 6. Temperature cross-sections by precipitation percentile

The companion to the paper's percentile figure, with one curve per date instead of one
per model. Each panel fixes precipitation at a percentile of the observed distribution
and draws the temperature response in 1961, 1980, 2000 and 2020.

Conventions follow the paper's figure. Every curve is **centred at the sample-mean
temperature** so it passes through zero there, and the panels therefore show growth
differences relative to the average temperature rather than absolute levels. That
centring is not only convention here: the dynamic model carries no additive year fixed
effects, so the level of `f` in any given year is confounded with the year effect and is
not interpretable on its own. Only the shape is.

The paper plots the same quantity in Δ ln(GDP); these panels are in percentage points,
consistent with the rest of this notebook. Divide by 100 to compare directly.

1960 is not in the sample, which starts in 1961, so the first curve is 1961 and the
remaining three sit on the round decades. Precipitation percentiles are taken over the
whole sample: the distribution is stable by decade (median 1072 to 1180 mm), so fixing
the cut levels once does not bias the comparison across dates.


```python
from plotly.subplots import make_subplots

####################### Percentile cross-section settings #######################
pct_node    = top_models[0]            # architecture to cut
percentiles = [10, 50, 90]             # one panel each; use [50] for a single panel
cut_years   = [1961, 1980, 2000, 2020] # one curve per date
n_temp      = 200                      # points along the temperature axis
year_cols   = ["#2f6f9f", "#5ba3a6", "#e08a3c", "#b23a48"]
#################################################################################

w_pct = read_weights(str(run_dir / "parameters" / f"{pct_node}.weights.h5"))
levels_mm = np.nanpercentile(Pv[np.isfinite(Pv)], percentiles)
Tcut = np.linspace(t_lo, t_hi, n_temp)
centre_idx = int(np.argmin(np.abs(Tcut - mean_temp)))

missing = [y for y in cut_years if y not in set(years)]
if missing:
    raise ValueError(f"years not in the sample {years[0]}-{years[-1]}: {missing}")

def cross_section(level_mm, year):
    """Centred temperature response in pp/yr at fixed precipitation and date."""
    j = int(np.where(years == year)[0][0])
    cols = [(Tcut - mean_temp) / std_temp,
            np.full_like(Tcut, (level_mm - mean_precip) / std_precip)]
    if bool(m.get("dynamic_model", False)):
        cols.append(np.full_like(Tcut, ((j + 1) - mean_t) / std_t))
    y = forward(w_pct, np.stack(cols, axis=-1))
    return (y - y[centre_idx]) * 100.0

titles = [f"{q}th percentile ({lvl:,.0f} mm)" for q, lvl in zip(percentiles, levels_mm)]
fig_pct = make_subplots(rows=1, cols=len(percentiles), shared_yaxes=True,
                        subplot_titles=titles, horizontal_spacing=0.04)

for col, lvl in enumerate(levels_mm, start=1):
    for k, yr in enumerate(cut_years):
        fig_pct.add_trace(
            go.Scatter(x=Tcut, y=cross_section(lvl, yr), mode="lines", name=str(yr),
                       legendgroup=str(yr), showlegend=(col == 1),
                       line=dict(color=year_cols[k % len(year_cols)], width=2.5)),
            row=1, col=col)
    fig_pct.add_hline(y=0, line=dict(color="black", width=1, dash="dot"), row=1, col=col)
    fig_pct.add_vline(x=mean_temp, line=dict(color="grey", width=1, dash="dot"),
                      row=1, col=col)
    fig_pct.update_xaxes(title_text="Temperature (°C)", row=1, col=col)

fig_pct.update_yaxes(title_text="Growth response (pp/yr, centred at mean T)",
                     row=1, col=1)
fig_pct.update_layout(
    title=dict(text=f"{run_dir.name}   {pct_node}", x=0.5, font=dict(size=13)),
    legend=dict(orientation="h", x=0.5, xanchor="center", y=-0.18,
                bgcolor="rgba(255,255,255,0.85)", bordercolor="black", borderwidth=1),
    margin=dict(l=60, r=20, b=80, t=60), font=dict(size=11), height=420)
fig_pct.show()

spread = pd.DataFrame(
    {f"{q}th": [cross_section(lvl, y).max() - cross_section(lvl, y).min()
                for y in cut_years]
     for q, lvl in zip(percentiles, levels_mm)}, index=cut_years)
spread.index.name = "year"
print("response range (pp/yr) by date and precipitation percentile:")
spread.round(2)
```

## 7. Animated surface, one frame per year

Press **Play** to sweep the years, or drag the slider to hold a single one. Each frame
shows that year's fitted `(T, P)` surface together with the country observations that
identify it.

**The observations are partialled out, not raw.** The network is fitted to growth after
the nuisance structure is projected out, so raw growth is not on the same scale as the
surface and overlaying it would mislead. Each point is plotted as

`y_it - (W gamma)_it`

where `W` holds the country fixed effects (plus year effects and country trends when the
run carries them) and `gamma` is recovered by exact OLS given the fitted climate net,
exactly as the training loss does it. The cell prints the mean gap between the two; it
should be zero to machine precision.

The z axis will not be centred on zero. The dynamic model carries no additive year fixed
effects, so `f` absorbs a common time level that is not separately identified. Leaving it
visible is more honest than silently recentring.


```python
import importlib.util

########################### Animation settings ###########################
anim_node   = top_models[2]      # architecture to animate
n_grid      = 70                 # grid points per axis (higher = smoother, bigger file)
precip_cap  = 3000.0             # mm, upper end of the precipitation axis
marker_size = 2
marker_col  = "#b23a48"
play_ms     = 1000                # Play button frame duration
step_ms     = 200                # slider step duration
camera      = dict(eye=dict(x=2.11, y=0.12, z=0.38))
##########################################################################

# WithinProjector is imported by path: the package __init__ chain pulls in tensorflow.
_spec = importlib.util.spec_from_file_location(
    "within_projection",
    ROOT / "src/models/helper_functions/global_model/within_projection.py")
_wp = importlib.util.module_from_spec(_spec)
sys.modules[_spec.name] = _wp
_spec.loader.exec_module(_wp)

dyn    = bool(m.get("dynamic_model", False))
ct     = bool(m.get("country_trends", False))
levels = str(m.get("target_mode", "growth")).lower() == "levels"

Yp = data.pivot(index="Year", columns="CountryCode",
                values="GDPCap" if levels else "GrowthWDI").reindex_like(Tp)
Yv = np.log(Yp.to_numpy(float)) if levels else Yp.to_numpy(float)
n_country = Tv.shape[1]

w_anim = read_weights(str(run_dir / "parameters" / f"{anim_node}.weights.h5"))

# --- observations, partialled out ---------------------------------------
proj = _wp.WithinProjector(np.isnan(Tv), country_trends=ct,
                           quadratic_trends=ct, include_time=not dyn)
_cols = [np.nan_to_num((Tv - mean_temp) / std_temp),
         np.nan_to_num((Pv - mean_precip) / std_precip)]
if dyn:
    _cols.append((np.arange(1, n_periods + 1)[:, None] - mean_t) / std_t
                 * np.ones((1, n_country)))
f_full = forward(w_anim, np.stack(_cols, axis=-1))

y_o = Yv[proj.t_arr, proj.n_arr]
f_o = f_full[proj.t_arr, proj.n_arr]
good = np.isfinite(y_o) & np.isfinite(f_o)
gamma = proj.recover_gamma(np.where(good, y_o - f_o, 0.0))
resid = np.where(good, y_o, np.nan) - proj.W @ gamma

obs = pd.DataFrame(dict(year=years[proj.t_arr],
                        T=Tv[proj.t_arr, proj.n_arr],
                        P=Pv[proj.t_arr, proj.n_arr] / 1000.0,
                        z=resid)).dropna()
print(f"observations {len(obs)} over {obs.year.nunique()} years   "
      f"mean gap (resid - fitted) = {np.nanmean(resid[good] - f_o[good]):+.2e}")

# --- surface per year ----------------------------------------------------
Tgrid = np.linspace(0.0, 30.0, n_grid)
Pgrid = np.linspace(float(np.nanmin(Pv)), precip_cap, n_grid)
TTa, PPa = np.meshgrid(Tgrid, Pgrid)

def _surf(j):
    cs = [(TTa - mean_temp) / std_temp, (PPa - mean_precip) / std_precip]
    if dyn:
        cs.append(np.full_like(TTa, ((j + 1) - mean_t) / std_t))
    return forward(w_anim, np.stack(cs, axis=-1))

Zs = [_surf(j) for j in range(n_periods)]
zlo = min(float(np.min(z)) for z in Zs + [obs.z.quantile(0.01)])
zhi = max(float(np.max(z)) for z in Zs + [obs.z.quantile(0.99)])
zpad = 0.05 * (zhi - zlo)

def _frame(j):
    o = obs[obs.year == years[j]]
    return [
        go.Surface(x=TTa, y=PPa / 1000.0, z=Zs[j], colorscale="Cividis",
                   cmin=zlo, cmax=zhi, opacity=0.9, showscale=False,
                   hoverinfo="skip", name="Model"),
        go.Scatter3d(x=o["T"], y=o["P"], z=o["z"], mode="markers",
                     name="Observations", showlegend=False,
                     marker=dict(size=marker_size, opacity=0.9, color=marker_col,
                                 line=dict(width=0.3)),
                     hovertemplate="T %{x:.1f} °C<br>P %{y:.2f} m"
                                   "<br>resid %{z:.3f}<extra></extra>"),
    ]

fig_anim = go.Figure(
    data=_frame(0),
    frames=[go.Frame(data=_frame(j), name=str(years[j])) for j in range(n_periods)])

fig_anim.update_layout(
    margin=dict(l=0, r=0, b=0, t=30),
    title=dict(text=f"{run_dir.name}   {anim_node}", x=0.5, font=dict(size=13)),
    scene=dict(
        xaxis=dict(title=dict(text="Temperature (°C)")),
        yaxis=dict(title=dict(text="Precipitation (m)")),
        zaxis=dict(title=dict(text="Δ ln(GDP per capita)" if levels else "Δ ln(Growth)"),
                   range=[zlo - zpad, zhi + zpad]),
        camera=camera),
    updatemenus=[dict(
        type="buttons", showactive=False, x=-0.01, xanchor="right", y=0.05,
        pad=dict(r=10, t=60),
        buttons=[
            dict(label="Play", method="animate",
                 args=[None, dict(frame=dict(duration=play_ms, redraw=True),
                                  transition=dict(duration=2),
                                  fromcurrent=True, mode="immediate")]),
            dict(label="Pause", method="animate",
                 args=[[None], dict(frame=dict(duration=0, redraw=False),
                                    mode="immediate", transition=dict(duration=0))])])],
    sliders=[dict(active=0, currentvalue=dict(prefix="Year: "), pad=dict(t=50),
                  steps=[dict(label=str(y), method="animate",
                              args=[[str(y)], dict(mode="immediate",
                                                   frame=dict(duration=step_ms, redraw=True),
                                                   transition=dict(duration=0))])
                         for y in years])])
fig_anim.show()
```

## 8. Initialisation stability

Sections 5 and 6 compare architectures. This section compares *initialisations of
the same architecture*, which is the only way to tell a genuine finding from a
local optimum the optimiser happened to land in.

Requires a run estimated with `use_diagnostics: true`, which writes
`diagnostics/<node>/initialisations.csv` and one weight snapshot per
initialisation. See `notes/2026-09-24/2026-09-24-convergence-diagnosis.md`.

Two things to read here:

1. `lr_final` confirms or refutes the annealing diagnosis directly. Before the
   fix, `EarlyStopping(patience=200)` fired before `ReduceLROnPlateau` could
   apply more than one cut, so `lr_final` should sit at 1e-3 or 3e-4 and never
   approach `lr_min`.
2. `BIC_spread` is the measured yardstick. The dynamic run's top two
   architectures are 9.65 BIC apart. If the spread *within* one architecture is
   comparable, that gap is not selecting anything.


```python
############################## User Input ##############################
stab_run = run_dir          # any run estimated with use_diagnostics: true
bic_tol  = 2.0              # inits within this many BIC of the best count as equally good
########################################################################

diag = stab_run / "diagnostics"
if not diag.is_dir():
    raise FileNotFoundError(
        f"{diag} does not exist. Re-estimate with use_diagnostics: true.")

frames = {}
for nd in sorted(diag.iterdir()):
    f = nd / "initialisations.csv"
    if f.is_file():
        frames[nd.name] = pd.read_csv(f).set_index("init").sort_index()

rows = []
for node, df in frames.items():
    b = df["BIC"].to_numpy(float)
    b = b[np.isfinite(b)]
    if b.size == 0:
        continue
    lrf = pd.to_numeric(df.get("lr_final"), errors="coerce")
    rows.append(dict(
        arch=node,
        n_inits=int(b.size),
        BIC_best=b.min(),
        BIC_spread=b.max() - b.min(),
        BIC_sd=b.std(ddof=1) if b.size > 1 else np.nan,
        n_within_tol=int((b <= b.min() + bic_tol).sum()),
        epochs_med=float(np.nanmedian(pd.to_numeric(df.get("epochs"), errors="coerce"))),
        lr_final_med=float(np.nanmedian(lrf)) if lrf is not None else np.nan,
        lr_reached_min=int((lrf <= 1.1e-7).sum()) if lrf is not None else -1,
    ))

stab = pd.DataFrame(rows).sort_values("BIC_best").reset_index(drop=True)
print(f"{stab_run.name}")
print(f"{len(stab)} architectures, {int(stab.n_inits.sum())} initialisations, "
      f"bic_tol = {bic_tol}\n")
stab.round(4)
```

**How to read the table.**

`lr_reached_min` counts initialisations whose learning rate actually annealed to
`lr_min`. Zero across the board means the schedule never ran and no fit reached a
stationary point, so nothing below it can be trusted. Values close to `n_inits`
mean the fix worked.

`BIC_spread` against the between-architecture gap decides whether selection is
meaningful. With `n = 10324`, 1 percent of MSE is `log(1.01) * n = 103` BIC, so a
spread of 10 BIC is about 0.1 percent of MSE.


```python
############################## User Input ##############################
stab_node = str(top_models[0])   # directory name under diagnostics/
n_temp_s  = 120
########################################################################

d  = stab_run / "diagnostics" / stab_node
df = frames[stab_node]

# Same grid, moments and centring as section 3, so these curves are directly
# comparable to the ones plotted there.
Tg_s   = np.linspace(t_lo, t_hi, n_temp_s)
TT, tt = np.meshgrid(Tg_s, np.arange(1, n_periods + 1), indexing="ij")
X_s    = np.stack([(TT - mean_temp) / std_temp,
                   np.full_like(TT, (precip_med - mean_precip) / std_precip),
                   (tt - mean_t) / std_t], axis=-1)

recs, curves = [], {}
for j in df.index:
    f = d / f"init_{j}.weights.h5"
    if not f.is_file():
        continue
    Gi   = forward(read_weights(str(f)), X_s)
    Gi   = (Gi - Gi.mean(axis=0, keepdims=True)) * 100.0
    gbar = Gi.mean(axis=1, keepdims=True)
    r    = Gi - gbar
    curves[j] = gbar[:, 0]
    recs.append(dict(
        init=int(j),
        BIC=float(df.loc[j, "BIC"]),
        dBIC=float(df.loc[j, "BIC"] - df["BIC"].min()),
        opt_T=float(Tg_s[np.argmax(gbar[:, 0])]),
        amp_pp=float(gbar.max() - gbar.min()),
        time_varying_pct=float((r**2).sum() / (Gi**2).sum() * 100),
        corr_first_last=float(np.corrcoef(Gi[:, 0], Gi[:, -1])[0, 1]),
    ))

init_stab = pd.DataFrame(recs).sort_values("BIC").reset_index(drop=True)
keep = init_stab[init_stab.dBIC <= bic_tol]

print(f"{stab_node}: {len(init_stab)} initialisations, "
      f"{len(keep)} within {bic_tol} BIC of the best\n")
print(f"  spread among the {len(keep)} equally good fits:")
for c in ("opt_T", "amp_pp", "time_varying_pct", "corr_first_last"):
    v = keep[c].to_numpy(float)
    print(f"    {c:>18}  min {v.min():9.3f}  max {v.max():9.3f}  spread {v.max()-v.min():9.3f}")

init_stab.round(3)
```

```python
fig_stab = go.Figure()
for j in init_stab.init:
    inside = bool((init_stab.init == j).any() and
                  float(init_stab.loc[init_stab.init == j, "dBIC"].iloc[0]) <= bic_tol)
    fig_stab.add_trace(go.Scatter(
        x=Tg_s, y=curves[j], mode="lines", name=f"init {j}",
        line=dict(width=1.6 if inside else 0.8,
                  color="#b23a48" if inside else "#b8b8b8"),
        opacity=0.85 if inside else 0.45, showlegend=False,
        hovertemplate=f"init {j}<br>%{{x:.1f}} C<br>%{{y:.2f}} pp<extra></extra>"))

fig_stab.update_layout(
    template="plotly_white", height=430,
    title=dict(text=f"{stab_node}: time-invariant response by initialisation "
                    f"(red = within {bic_tol} BIC of best, grey = worse)", x=0.5,
               font=dict(size=13)),
    xaxis_title="Temperature (C)",
    yaxis_title="Response (pp/yr, centred)")
fig_stab.show()
```
